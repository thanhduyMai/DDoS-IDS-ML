```python

from pyspark.sql import SparkSession
spark = SparkSession.builder \
    .appName("BigData_CICDDoS") \
    .config("spark.driver.memory", "8g") \
    .config("spark.executor.memory", "8g") \
    .config("spark.executor.cores", "4") \
    .getOrCreate()
```

```python

# Thư mục dữ liệu
csv_folder = "CIC_DDoS_CSVs/01-12"
parquet_folder = "CIC_DDoS_Parquet"

# Tạo thư mục lưu file Parquet nếu chưa có
os.makedirs(parquet_folder, exist_ok=True)

# Danh sách file CSV (trừ TFTP)
csv_files = [
    "DrDoS_SNMP.csv", "DrDoS_DNS.csv", "DrDoS_MSSQL.csv",
    "DrDoS_NetBIOS.csv", "DrDoS_UDP.csv", "DrDoS_SSDP.csv",
    "DrDoS_LDAP.csv", "DrDoS_NTP.csv", "Syn.csv", "UDPLag.csv"
]

# Xử lý từng file CSV
for csv_file in csv_files:
    csv_path = os.path.join(csv_folder, csv_file)
    parquet_path = os.path.join(parquet_folder, csv_file.replace(".csv", ""))  # Không cần .parquet

    print(f" Đang xử lý {csv_file} ...")

    # Đọc CSV bằng Spark
    df = spark.read.csv(csv_path, header=True, inferSchema=True)

    # Ghi thành Parquet (fix lỗi thư mục)
    df.repartition(10).write.mode("overwrite").parquet(parquet_path)

    print(f" Đã lưu {parquet_path}")

print("Hoàn thành chuyển đổi tất cả file CSV !")
```

```python
import os

parquet_folder = "CIC_Reflection_DDoS_Parquet"
all_parquet_files = [os.path.join(parquet_folder, d) for d in os.listdir(parquet_folder) if os.path.isdir(os.path.join(parquet_folder, d))]

# Đọc tất cả file Parquet vào DataFrame Spark
df = spark.read.parquet(*all_parquet_files)

# Kiểm tra số dòng sau khi nối
print("Tổng số dòng sau khi nối:", df.count())

# Hiển thị 5 dòng đầu tiên
df.show(5)
```

```python
from pyspark.sql.functions import col

# Hàm làm sạch tên cột
def clean_column_name(name):
    return name.strip().replace(" ", "_").replace(".", "_")

# Xử lý tên cột
new_column_names = [clean_column_name(c) for c in df.columns]

# Đổi tên các cột trong DataFrame
df = df.toDF(*new_column_names)
```

```python
from pyspark.sql.functions import col

df = df.withColumnRenamed("Flow_Bytes/s", "Flow_Bytes_per_s")
df = df.withColumnRenamed("Flow_Packets/s", "Flow_Packets_per_s")
```

```python
columns_to_keep = [
    'Source_IP', 'Source_Port', 'Destination_IP', 'Destination_Port', 'Protocol', 'Flow_Duration',
    'Total_Fwd_Packets', 'Total_Backward_Packets', 'Total_Length_of_Fwd_Packets', 'Total_Length_of_Bwd_Packets',
    'Fwd_Packet_Length_Max', 'Fwd_Packet_Length_Mean', 'Fwd_Packet_Length_Std',
    'Bwd_Packet_Length_Max', 'Bwd_Packet_Length_Mean', 'Bwd_Packet_Length_Std',
    'Flow_Bytes_per_s', 'Flow_Packets_per_s',
    'Flow_IAT_Mean', 'Flow_IAT_Std', 'Flow_IAT_Max', 'Flow_IAT_Min',
    'Fwd_IAT_Total', 'Fwd_IAT_Mean', 'Fwd_IAT_Std', 'Fwd_IAT_Max', 'Fwd_IAT_Min',
    'Bwd_IAT_Total', 'Bwd_IAT_Mean', 'Bwd_IAT_Std', 'Bwd_IAT_Max', 'Bwd_IAT_Min',
    'FIN_Flag_Count', 'SYN_Flag_Count', 'RST_Flag_Count', 'PSH_Flag_Count', 'ACK_Flag_Count',
    'Init_Win_bytes_forward', 'Init_Win_bytes_backward', 'act_data_pkt_fwd', 'min_seg_size_forward',
    'Active_Mean', 'Active_Std', 'Active_Max', 'Active_Min',
    'Idle_Mean', 'Idle_Std', 'Idle_Max', 'Idle_Min',
    'Label'
]

df = df.select(columns_to_keep)
```

```python
from pyspark.sql.functions import col, isnan, when, count

df.select([
    count(
        when(
            col(c).isNull() | isnan(col(c)) | (col(c) == float('inf')) | (col(c) == float('-inf')),
            c
        )
    ).alias(c) for c in df.columns
]).show()
```

```python
from pyspark.sql.functions import when, col

# 1. Tính median cho từng cột
median_flow_bytes = df.approxQuantile("Flow_Bytes_per_s", [0.5], 0.001)[0]
median_flow_packets = df.approxQuantile("Flow_Packets_per_s", [0.5], 0.001)[0]

# 2. Thay Inf bằng median tương ứng
df = df.withColumn(
    "Flow_Bytes_per_s",
    when(
        (col("Flow_Bytes_per_s") == float('inf')) | (col("Flow_Bytes_per_s") == float('-inf')),
        median_flow_bytes
    ).otherwise(col("Flow_Bytes_per_s"))
)

df = df.withColumn(
    "Flow_Packets_per_s",
    when(
        (col("Flow_Packets_per_s") == float('inf')) | (col("Flow_Packets_per_s") == float('-inf')),
        median_flow_packets
    ).otherwise(col("Flow_Packets_per_s"))
)
```

```python
df.select([
    count(
        when(
            col(c).isNull() | isnan(col(c)) | (col(c) == float('inf')) | (col(c) == float('-inf')),
            c
        )
    ).alias(c) for c in df.columns
]).show()
```

```python
from pyspark.sql.functions import col, isnan, when, count

df.select(
    count(when(col('Flow_Bytes_per_s').isNull(), True)).alias('null_count'),
    count(when(isnan(col('Flow_Bytes_per_s')), True)).alias('nan_count'),
    count(when(col('Flow_Bytes_per_s') == float("inf"), True)).alias('inf_count'),
    count(when(col('Flow_Bytes_per_s') == float("-inf"), True)).alias('ninf_count')
).show()
```

```python
from pyspark.sql.functions import col, when, lit

# Ví dụ thay NULL trong cột Flow_Bytes_per_s
median_value = df.approxQuantile('Flow_Bytes_per_s', [0.5], 0.001)[0]

df = df.withColumn(
    'Flow_Bytes_per_s',
    when(col('Flow_Bytes_per_s').isNull(), lit(median_value))
    .otherwise(col('Flow_Bytes_per_s'))
)
```

```python
df.select([
    count(
        when(
            col(c).isNull() | isnan(col(c)) | (col(c) == float('inf')) | (col(c) == float('-inf')),
            c
        )
    ).alias(c) for c in df.columns
]).show()
```

# **Quá trình huấn luyện mô hình**

## **Data preparation**

Sau khi đã có những khám phá về dataset , hiểu ý nghĩa của các feature , chúng ta đến với phần chuẩn bị dữ liệu cho mô hình phân loại - mô hình quan trọng trong một detection system


### **Feature Selection**


```python
cols_to_drop = [
    'Source_IP', 'Destination_IP', 'Source_Port', 'Destination_Port'
]
df = df.drop(*cols_to_drop)
df.show(5, truncate=False)
```

### **Pipeline input mô hình với Spark**

```python
from pyspark.ml import Pipeline
from pyspark.ml.feature import StringIndexer, VectorAssembler, StandardScaler

# 1. Encode Label
label_indexer = StringIndexer(inputCol="Label", outputCol="Label_Index", handleInvalid="keep")

# 2. Cột số cần scale (đã bao gồm Protocol vì nó là số hợp lệ)
numeric_cols = [
    'Protocol', 'Total_Fwd_Packets', 'Total_Backward_Packets', 'Total_Length_of_Fwd_Packets',
    'Total_Length_of_Bwd_Packets', 'Fwd_Packet_Length_Max', 'Fwd_Packet_Length_Mean',
    'Fwd_Packet_Length_Std', 'Bwd_Packet_Length_Max', 'Bwd_Packet_Length_Mean',
    'Bwd_Packet_Length_Std', 'Flow_Bytes_per_s', 'Flow_Packets_per_s', 'Flow_IAT_Mean',
    'Flow_IAT_Std', 'Flow_IAT_Max', 'Flow_IAT_Min', 'Fwd_IAT_Total', 'Fwd_IAT_Mean',
    'Fwd_IAT_Std', 'Fwd_IAT_Max', 'Fwd_IAT_Min', 'Bwd_IAT_Total', 'Bwd_IAT_Mean',
    'Bwd_IAT_Std', 'Bwd_IAT_Max', 'Bwd_IAT_Min', 'FIN_Flag_Count', 'SYN_Flag_Count',
    'RST_Flag_Count', 'PSH_Flag_Count', 'ACK_Flag_Count', 'Init_Win_bytes_forward',
    'Init_Win_bytes_backward', 'act_data_pkt_fwd', 'min_seg_size_forward', 'Active_Mean',
    'Active_Std', 'Active_Max', 'Active_Min', 'Idle_Mean', 'Idle_Std', 'Idle_Max', 'Idle_Min', 'Flow_Duration'
]

# 3. Assemble features
assembler = VectorAssembler(inputCols=numeric_cols, outputCol="features", handleInvalid="keep")


# 4. Pipeline
pipeline = Pipeline(stages=[
    label_indexer,
    assembler
])

# 6. Fit & transform
pipeline_model = pipeline.fit(df)
df_transformed = pipeline_model.transform(df)

# 7. Kết quả
df_final = df_transformed.select("features", "Label_Index")
```

### **Áp dụng mô hình**

#### **Mô hình Random Forests**

##### **Tách train/test & Tối ưu tham số**

```python
from pyspark.ml.classification import RandomForestClassifier
from pyspark.ml.evaluation import MulticlassClassificationEvaluator
from pyspark.ml.tuning import ParamGridBuilder, TrainValidationSplit

# 1. Chia train/test
train_df, test_df = df_final.randomSplit([0.8, 0.2], seed=42)

# 2. Định nghĩa Random Forest model
rf = RandomForestClassifier(
    labelCol="Label_Index",
    featuresCol="features",
    seed=42
)

# 3. Tạo param grid (rất nhẹ)
paramGrid = (ParamGridBuilder()
             .addGrid(rf.numTrees, [50, 100])
             .addGrid(rf.maxDepth, [5, 7])
             .build())

# 4. Định nghĩa evaluator (Recall)
evaluator = MulticlassClassificationEvaluator(
    labelCol="Label_Index",
    predictionCol="prediction",
    metricName="weightedRecall"   # 🎯 Dùng weightedRecall
)

# 5. TrainValidationSplit
tvs = TrainValidationSplit(
    estimator=rf,
    estimatorParamMaps=paramGrid,
    evaluator=evaluator,
    trainRatio=0.8,
    parallelism=8
)

# 6. Train mô hình
tvs_model = tvs.fit(train_df)

# 7. Dự đoán trên test set
predictions = tvs_model.transform(test_df)

# 8. Đánh giá
recall = evaluator.evaluate(predictions)
print(f"Test Recall = {recall:.4f}")

# 9. In tham số của best model
best_model = tvs_model.bestModel
print("Best numTrees:", best_model.getNumTrees)
print("Best maxDepth:", best_model.getOrDefault('maxDepth'))
```

##### **Áp dụng mô hình RF**

```python
from pyspark.ml.classification import RandomForestClassifier
from pyspark.ml.evaluation import MulticlassClassificationEvaluator

# 1. Chia dữ liệu train/test
train_df, test_df = df_final.randomSplit([0.8, 0.2], seed=42)

# 2. Tạo mô hình Random Forest
rf = RandomForestClassifier(
    labelCol="Label_Index",
    featuresCol="features",
    numTrees=50,        # Số cây vừa đủ (không quá nặng)
    maxDepth=7,         # Giới hạn độ sâu cây
    seed=42
)

# 3. Train mô hình
rf_model = rf.fit(train_df)

# 4. Dự đoán trên test set
predictions = rf_model.transform(test_df)
```

##### **Đánh giá mô hình**

```python
from pyspark.ml.evaluation import MulticlassClassificationEvaluator

# 5. Đánh giá hiệu suất
evaluator_f1 = MulticlassClassificationEvaluator(
    labelCol="Label_Index",
    predictionCol="prediction",
    metricName="f1"
)

evaluator_accuracy = MulticlassClassificationEvaluator(
    labelCol="Label_Index",
    predictionCol="prediction",
    metricName="accuracy"
)

evaluator_precision = MulticlassClassificationEvaluator(
    labelCol="Label_Index",
    predictionCol="prediction",
    metricName="weightedPrecision"
)

evaluator_recall = MulticlassClassificationEvaluator(
    labelCol="Label_Index",
    predictionCol="prediction",
    metricName="weightedRecall"
)

# Tính các chỉ số
f1_score = evaluator_f1.evaluate(predictions)
accuracy = evaluator_accuracy.evaluate(predictions)
precision = evaluator_precision.evaluate(predictions)
recall = evaluator_recall.evaluate(predictions)

# In kết quả
print(f"✅ F1-score: {f1_score:.4f}")
print(f"✅ Accuracy: {accuracy:.4f}")
print(f"✅ Precision: {precision:.4f}")
print(f"✅ Recall: {recall:.4f}")
```

#### **Mô hình MLP**

##### **Tối ưu tham số**

```python
from pyspark.ml.classification import MultilayerPerceptronClassifier
from pyspark.ml.tuning import ParamGridBuilder, TrainValidationSplit
from pyspark.ml.evaluation import MulticlassClassificationEvaluator

# 1. Lấy số chiều input feature và số lớp output
input_dim = train_df.select("features").first()[0].size  # số chiều input
output_dim = train_df.select("Label_Index").distinct().count()  # số lớp

# 2. Định nghĩa MLP model
mlp = MultilayerPerceptronClassifier(
    labelCol="Label_Index",
    featuresCol="features",
    maxIter=100,
    layers=[input_dim, 16, 8, output_dim],  # bạn có thể điều chỉnh
    blockSize=128,
    seed=42
)

# 3. Tạo param grid để tuning
paramGrid = (ParamGridBuilder()
             .addGrid(mlp.maxIter, [100, 150])
             .addGrid(mlp.stepSize, [0.01, 0.1])
             .build())

# 4. Định nghĩa evaluator (Recall)
evaluator = MulticlassClassificationEvaluator(
    labelCol="Label_Index",
    predictionCol="prediction",
    metricName="weightedRecall"
)

# 5. TrainValidationSplit
tvs = TrainValidationSplit(
    estimator=mlp,
    estimatorParamMaps=paramGrid,
    evaluator=evaluator,
    trainRatio=0.8,
    parallelism=8
)

# 6. Train mô hình
tvs_model = tvs.fit(train_df)

# 7. Dự đoán trên test set
predictions = tvs_model.transform(test_df)

# 8. Đánh giá
recall = evaluator.evaluate(predictions)
print(f"Test Recall = {recall:.4f}")

# 9. In cấu trúc mô hình tốt nhất
best_model = tvs_model.bestModel
print("Best layers:", best_model.getLayers())
print("Best stepSize:", best_model.getStepSize())
```
